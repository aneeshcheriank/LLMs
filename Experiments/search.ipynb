```python
from ddgs import DDGS
import trafilatura

def search_url(query):
    with DDGS() as ddgs:
        results = [r['href'] for r in ddgs.text(
            query, max_results=5
        )]
    
    if results:
        return results
    return None

def cleaned_text(urls):
    results = []
    for n, url in enumerate(urls):
        downloaded = trafilatura.fetch_url(url)
        # remove ads, navbars, and SEO Junk
        clean_text = trafilatura.extract(
            downloaded, include_comments=False, output_format="markdown"
        )
        results.append({
            'url': url,
            'text': clean_text
        })
    return results

def search(query):
    urls = search_url(query)
    if urls is not None:
        result = cleaned_text(urls)
        return result
    return None
```

```python
result = search("what is the grownth projections of different sectors in Germany and Europien union")
```
