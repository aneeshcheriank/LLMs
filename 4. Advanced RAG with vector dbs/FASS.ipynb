- FAISS provides diffeent index types 
- each index balance speed, memory and accuracy differently

#### Flat
- compares the distance (Eculidien, dot product) between query and every embedding using brute force search
- gives back k nearest neighbors from nearest to furthest
- slow

#### IVF (Inverted File Index)
- Clustering vectors using clustering like K-means
- forming cells, each cell contains vector closest to centroid
- search is limitted to vectos in teh nearest cell
- Fast
- Accuracy is limitted since the point may be near to a vector that may be in another cell

#### Locality-Sensitive Hashing (LSH)
- Uses hash functions to goup similar vectors 
- Fast and memory efficient
- Best for high-dimentional sparse data
- not the fastest or most

- arrange vectos to layers
- top layer has only few vectors and sparse
- it helps the query to reach the region fast

- Fast and accurate for large dataset
