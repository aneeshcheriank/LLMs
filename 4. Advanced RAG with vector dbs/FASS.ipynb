FAISS
- Facebook AI Similarity Search
- for fast vector search
- run on a single machine (CPU or )
- has no builit in database or server
- Great for custom high performance systems

Chroma db
- Vector db for AI appliations
- Stores vectors and Metadata
- Can run locally or as a 

### FAISS vs Chroma db
|Features|FAISS|Chroma DB|
|---|---|---|
|Type|Library|Database|
|Deployment|Local, single node| Local, single node, or distributed|
|Indexing Options|Flat, IVF, LSH, HNSW|HNSW only|
|Metadata support|None|Storage, Filtering|
|LangChain/ LlamaIndex support|Yes|Yes|

- FAISS provides diffeent index types 
- each index balance speed, memory and accuracy differently
