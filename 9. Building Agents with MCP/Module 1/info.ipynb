## What is MCP
- model context protocol
- MCP Host has one or more MCP client (chat, IDE)
- MCP Host connected to MCP Server (can connect to multiple severs)
- the connection is using MCP protocol
- whenever, MPC Host need a tool it will connect to a Sever
    - MCP sever can connect to variety of services
        - DBs, APIs, Code



### MCP - LLM interactions
- MCP host get a question from the user
- check in the MCP server to get the cateloge of the tools
- MCP host send the quetions along with the available tools to LLM
- LLM replay with which tool to call back to MCP host
- MCP host will decide which server to call
- the MCP server will execute the tool get the result back to LLM through MCP client

### Introduction
- MCP is an open protocol to build MCP servers using multiple programming languages
- Standardizes connction between AI applications, LLMs, and external data
- LLMs (AI agents) have 2 main needs
    - contextual data
    - tools and capabilites
- AI applciations can query MCP servers to determine what capabilites available

### Importance of standardization
- MCP behaves like a universal layer
    - built tools that interact with any resouce in a uniform manner



### Key Benifits of MCP
- Standardized integration


- simple architecture


- Interoperability


- Security


- Minimized AI hallucinations
    - LLMs use internal training data to provide best possible responses
    - so when the information is not available it gives a postive response called AI hallucination
    - MCP provides upto date data from external data souces or various tools to extrat accurate upto-date data
