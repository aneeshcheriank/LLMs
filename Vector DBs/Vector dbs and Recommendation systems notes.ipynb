## Database operations in Chroma db
- Create and Modify collections
- add, update, and delete documents

````python
## Create collections
import chromadb
from chromadb.utils import embedding_functions

ef = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-MiniLM-L6-v2"
)

client = chromadb.Client()

collection = client.create_collection(
    name="my_collection",
    metadata={"description": "A collection for storing user data"},
    configuration={
        "hnsw":{
            "space":"cosine"
        },
        "embedding_function": ef
    }
)
````

- Connecting to existing connection

```python
client.get_collection(name="my_connection")
```

## Modifying a collection
- modification can't change
    - Embedding model
    - Distance metric
- in this case need a collection clone
    - the operation is costly if a substantial amount of data is stored in the repository

````python
collection.modify(
    name="test2", 
    metadata={"key": "value"}
)

print(f"COllection metadata: {collection.metadata}")
````

## Adding docuemnts to colelction

````python
collection.add(
    documents=[
        "This is a document about LangChain",
        "This is a docuemnt about LlamaIndex"
    ],
    metadatas=[
        {"source": "langchain.com", "version": "0.2"},
        {"source": "llamaindex.ai", "version": "0.12"}
    ],
    ids=["id1", "id2"]
)
````

- get colletion items
    - `collection.get()`
    - returns a python dictionary
    - to return embeddings
        - `collecion.get(include=["embeddings"])`
    - to get individual collection
        - `collection.get("id1")`

## Updating data in collections
- selecting document by the id
- updating metadata and document

````python
collection.update(
    ids=["id1"],
    metadatas=[{"source":"langchain.com", "version":"0.3"}],
    documents=["This is an upated document about LangChain"]
)
````

## Delete
- either pass Ids
- where filter
- or a combination of Ids and filters

````python
collection.delete(
    ids=["id1", "id2"],
    where={"source": "llamaindex.ai"}
)
````
