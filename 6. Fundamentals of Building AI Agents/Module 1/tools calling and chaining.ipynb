### 4 criteria framework for using AI agents
- before using agents you have to ask these questions

#### is the task ambiguous or predictable?
- use **agents** when task is ambiguous
    - decision path is unclear or can't mapped in advance
    - task involves exploration, troubleshooting, or creativity

- use **wrokflows** when the task is predictable
    - when you can defien all rules and outcomes
    - the process follows a clear, repeatable structure

#### is the value of the taks worth the cost?
- agents are more expensive due to exploaration overhead
- can consume 10 to 100x more toakens thatn a workflow

|Scenario|Recommendation|
|---|---|
|Strategic planning with high ROI|Use an agent|
|Basic customer support task|Use workflow|

#### does the agent meet minimum capabilites?
- test the agents on 3 - 5 key skills
- you have to define these key skills
- for example
    - A research agent must identify, filter, and summarize credible sources
    - A coding agent must write, fix, and validate code snippets
    - A customer support agent must classify issues, resolve common queries, and escalate complex cases appropriately
    - A data analysis agent must clean datasets, detect anomalies, and summarize key trends
- if the agent fails these tests, **scale back or redesign the agents**

#### what happend if the agent makes mistakes?
- Evaluate the answer to these questions
    - Can you catch and correct errors quickly? If so, then using an agent might be appropriate.

    - What's the risk if something is missed? Does the consequence of missing the answer affect the customer’s or organization’s well-being or safety?

    - Does the agent include built-in correction or validation tools?

    - Use agents when risk is manageable or reversible.

### When not to use agents
- **High-volume, low-margin tasks**, such as basic chat support
- **Real-time applications**, such as instant fraud detection
- **Zero-error systems**, including medical or security decisions
- **Heavily reguralted industries need deterministic outcomes**

### Guidline for managing risk when deploying agents
- some helpful tips for effective agent architecture

#### Priotitze key elements of effective agent architecture
- keep the agent architecture sikmple
- 3 key components
    - Environment
    - Tools: tools agents used to act or observe
    - System prompt: the rules and behaviours that guide the agent's operation
- start simple with agents actions. Add task complexity only after confirming  the agent's reliable performance

#### Assessing your deployment plan
|if the risk Level is...|Implement the following response stragegy|
|---|---|
|High-stakes and difficult to notice|Use Human review and multiple validation layers|
|High-stakes and visible|Add automated checks and oversight mechanisms|
|Low-stakes|Monitor the agent's action after feedback and lightweight validation|

- Best practices
    - Start with read-only access to tools and systems
    - Add human approvals for critical steps
    - Use staged deployments with monitoring
    - Enable comprehensive logging

### Implementing agent responsibly
- when implementing agents, consider a phased deployment plan

    1. **Validate the Proof of Concept:** Try low-risk, reversible tasks
    2. **Implement a Pilot Program:** Test the agent using moderate-risk tasks under supervision
    3. **Production Scaling:** Expand use of the agent only after demonstrating its safety and performance
