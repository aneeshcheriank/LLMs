## AI system design
- determining factors
    - task complexity
    - adaptiability needs
    - operational requirements

- 3 Main types (AI system design)
    - **Single LLM**
    - **Structured workflows**
    - **Autonomous AI agents**

### Single LLM


- Key characteristics
    - stateless processing: no retention of information or context across interactions.
    - direct input-output flows: strightforward request-response mechanism
    - Predefined tasks: Suitable only for clearly defined single-step actions

- Best suited for sell defined tasks required no memory
- examples
    - text summarization
    - sentiment classification
    - information extraction
    - translation

- Advantages
    - Speed and simplicity
    - Deterministic output: same intput -> same output
    - Low cost: minimal compute and orchestration overhead

- Limitations
    - No adaptability: can't handle context or dynamic decision-making
    - No memory: each input processed independently

### Sturctured workflows
- orchestrate LLM and tool calls through explicit, deterministic code paths.
- ideal for repetitive, multi-step, or compliance-heavy tasks
- Consider processing insurance claims, where each document is scanned, information is extracted, validated, and stored. Each step must follow a precise, predictable order, making structured workflows ideal.



- Key characterestics
    - **Deterministic execution:** Inputs produce consistent outputs.
    - **Explicit control flow:** All steps and decisions are predefined.
    - **Predefined tool chains:** Tool use is fixed and transparent.

- Uses
    - Repetitive, multi-step tasks with clear logic and minimal ambiguity
    - Regulatory or compliance-driven applications
    - Scenarios requiring consistency, traceability, and auditability

- Examples
    - Document and data pipelines (Optical Character Recognition (OCR) → extraction → validation → storage)
    - Batch report generation
    - Financial and healthcare transaction processing

- Advantages
    - Predictable and reliable: Easy to monitor, debug, and audit
    - Cost-efficient: No unnecessary exploration
    - Compliance-ready: Supports versioning, error handling, and audit trails

- Limitations
    - Rigidity: Difficulty adapting to new or ambiguous scenarios
    - Development overhead: The necessity to code each exception or variant

### Autonomous agents
- LLMs plans sequence of actions when conditions change
- choose tools and workflows on real-time context and feedback
- example: AI vaccation planner



- Capabilities
    - Dynamic planning: Decomposes goals and adjusts steps as needed
    - Contextual awareness: Remembers past steps and adapts to user and environment feedback
    - Tool orchestration: Selects tools and changes strategies dynamically

- Uses
    - Complex, open-ended tasks with unclear solution paths
    - Scenarios requiring real-time adaptation and reasoning
    - Environments with high variability or need for personalization

- Limitations
    - Unpredictable outcomes: Requires robust monitoring and safeguards
    - Higher complexity and cost: More difficult to debug and guarantee compliance

### 4 criteria framework for using AI agents
- before using agents you have to ask these questions

#### is the task ambiguous or predictable?
- use **agents** when task is ambiguous
    - decision path is unclear or can't mapped in advance
    - task involves exploration, troubleshooting, or creativity

- use **wrokflows** when the task is predictable
    - when you can defien all rules and outcomes
    - the process follows a clear, repeatable structure

#### is the value of the taks worth the cost?
- agents are more expensive due to exploaration overhead
- can consume 10 to 100x more toakens thatn a workflow

|Scenario|Recommendation|
|---|---|
|Strategic planning with high ROI|Use an agent|
|Basic customer support task|Use workflow|

#### does the agent meet minimum capabilites?
- test the agents on 3 - 5 key skills
- you have to define these key skills
- for example
    - A research agent must identify, filter, and summarize credible sources
    - A coding agent must write, fix, and validate code snippets
    - A customer support agent must classify issues, resolve common queries, and escalate complex cases appropriately
    - A data analysis agent must clean datasets, detect anomalies, and summarize key trends
- if the agent fails these tests, **scale back or redesign the agents**

#### what happend if the agent makes mistakes?
- Evaluate the answer to these questions
    - Can you catch and correct errors quickly? If so, then using an agent might be appropriate.

    - What's the risk if something is missed? Does the consequence of missing the answer affect the customer’s or organization’s well-being or safety?

    - Does the agent include built-in correction or validation tools?

    - Use agents when risk is manageable or reversible.

## Tools, Agents and Functiosn calling in LangChain

- Objectives
    - LLMs external world interaction with **tool calling** or **function calling**

### Tools
- Functions made available to LLMs
- a tool in LangChain has several **important components** that forms its schema
    - **Name**: a unique identifier for the tool
    - **Description**: a brief explanation of the tools purpose
    - **Parameters**: the inputs to the tool to function correctly
 - the schema allows the LLM to understand when and how to use the tool effectively

 ### Tool calling
 - LLM is not explicitly calling the tool
 - it generate a response which tool to call and with what parameters
 - the response (tool call) in JSON format
 - the external system then interpret the output, execute the call, and retrieve the result
 - the result fed back to the LLM, which then generate a comprhensive response

 ### Funtion calling vs Tool calling
 - same concept, with different terminology
- Function calling is the term popularized by OpenAI in their API documentation
- Tool calling is a more general industry term used by multiple AI providers, including Anthropic, and broadly in frameworks like LangChain
- Both describe the same capability: enabling an LLM to request specific external functions to be executed with structured parameters

### Tools in LangChain
- tools are utiliteis designed to be called by models
- these tools perform a specific action like serching web
- a **toolkit** is a collection of related tools designed to work together 

### ways to initalize and use tools
- several methods to initialize and use tools
    1. **Using Built-in Tools:**
        - Build-in tools for common tasks
        - example: WikipediaQueryRun used to fetch data from Wikipedia
    2. **Loading Tools with load_tools:**
        - load_tools function to load multiple tools conveniently.
        - to load tools like wikiepedia, serpapi, llm-math, etc
    3. **Cerating Custom Tools:**
        - can create custom tools
        - Tool class or @tool dectorator
    4. **Tools as OpenAI Functions:**
        - LangChain tools can be converted to OpenAI fucntion using the convertto_openai_function utility.
        - this allows you to use LangChain tools with OpenAI#s function calling API
        - bind_functions or bind_tools mentod to automaically convert and bind tools to OpenAI chat model
    - LangChain supports a wide range of external tools
    - Some popular categories include
        - Wikipedia: summaries and information from wikipedia articles
        - Search engines: Integrate seaerch engine like Bing, Google, and DuckDuckGo to fetch real-time search results
        - APIs: Access various APIs

### Agents
- Agents are decision making system powered by LLMs that can reason, use tools, access memory, and take actions to complete tasks.
- Agent is a high level orchestration system. It got supporting components like tools, memory, and execution framework.

### Architecture of an AI agent in LangChain


### Agents in LangChain
- several methods to create agents in LangChain

    1. **Using built-in agent types**
        - LangChain offers predefined agent types such as **zero-shot-react-description** and **chat-zero-shot-react-description**. These are useful for simple reasoning tasks where the model decides which tool to use based on the description alone.

    2. **Creating agents with OpenAI fucntions**
        - LangChain supports agents that can leverage OpenAI's **function-calling capabilities**. You can create such agents using create_openai_functions_agent, which makes it easy to interact with structured tools in a safe and predictable way.

    3. **Building agents with LangGraph**
        - LangGraph is a low-level orchestration library that provides more flexibility and control. Using the high-level **create_react_agent** method, you can create agents capable of complex reasoning, action chaining, memory integration, and real-time streaming.

    4. **Executing agents with AgentExecutor**
        - Agents are executed using AgentExecutor, which manages the loop of calling the LLM, processing tool outputs, and determining when the final response is ready. This ensures the agent can dynamically react to intermediate results during its reasoning process.

    5. **Adding Memory**
        - By using components like MemorySaver, agents can maintain context across multiple interactions. This allows for personalized and context-aware responses, where an agent can remember things like user preferences, past questions, or conversation history.

### Summary
1. **Tool calling fundamentals**: Tool calling allows LLMs to generate structured requests (not actual execution) that specify which external tool to use and with what parameters, enabling access to real-time or external data.
2. **Components of a tool**: A tool in LangChain includes a name, description, and parameters. This schema helps the model understand how and when to use each tool.
3. **Agents vs. tools**: Agents are advanced systems powered by LLMs that can make decisions, call tools, access memory, and take actions, enabling dynamic task completion beyond text generation.
4. **Workflow of tool invocation**: When a question requires external data, the LLM outputs a structured tool call. An external system executes the tool, retrieves results, and feeds them back to the LLM for a complete answer.
5. **Function calling vs. tool calling**: These terms are technically identical. "Function calling" is OpenAI's terminology; "tool calling" is used more broadly in LangChain and other frameworks.

### Zero-short ReAct agent
- uses zero-short resoning
- solve task it hasn't seen before by breaking the task step by step
- useful for simple and well structured tasks

````python
# Zero short react agent

# create a function to add numbers
def add_numners(inputs:str) -> dict:
    """
    Tool: add_numbers
    Description: Adds numeric values extracted from a string input.
    Input: str (e.g., "Add 10, 20, and 30")
    Output: dict (e.g., {"result": 60})
    """

    numbers = [int(num) for num in inputs.replace(",", "").split() if num.isdigit()]
    result = sum(numbers)
    return {"result": result}

# wrap the function in a tool object
# make it available to the agent
from langchain.agents import Tool

add_tool=Tool(
    name="AddTool",
    func=add_numners,
    description="A tool to add numbers from a string input."
)

# create a zero-shot react agent
from langchain.agents import initialize_agent

agent = initialize_agent(
    tools=[add_tool],
    llm = llm,
    agent = "zero-shot-react-description",
    verbose = True,
    handle_parsing_errors = True # agent to recover if the llm input is slightly off
)

# to run the agent
response = agent.run(
    "In 2023, the US GDP was approximately $27.72 trillion, while Canada's was around $2.14 trillion and Mexico's was about $1,.79 trillion, what is the total?"
)
````

- `.run()` method will not work with some agents
- use `.invoke()` method insted
    - useful for debugging 
    - working with more complex agents

#### Structued-chat-zero-short-react-description agent
- when selecting an agent, the tool format, input and return type matter
- example
    - **zero-shot-react-description** agent expect tools to accept and return plain text
    - **structured-chat-zero-shot-react-descrption** agent support strucuted tools and allow typed inputs and structured inputs


````python
from typing import List

@tool
def add_numbers_with_options(numbers: List[float], absolute:bool = False) -> float:
    """
    Adds a list of numbers provided
    """
    if absolute:
        numbers = [abs(n) for n in numbers]
    return sum(numbers)

agent = initialize_agent(
    tools=[add_numbers_with_options],
    llm = llm,
    agent = "structured-chat-zero-shot-react-description",
    verbose = True,
    handle_parsing_errors = True # agent to recover if the llm input is slightly off
)

response = agent.invoke({
    "input": "Add -10, 20, and -30 and give me the absolute value of the result."
})
````

## Build a Custom Math Toolkit Agent with LangChain
- Objective
    - create agents using LangChain and LangGraph
    - Build toolkit with tools for 
        - addition
        - subtraction
        - multiplication
        - division
    - Guide the agents with a custom prompt and run complex queries using invoke
    - Integrate tools and build agent that both calculate and retieve information

- LangGraph
    - preferred approach over 
    - create_react_agent
        - works with just one type
        - but can customize the agent
            - more control over the agents behaviour and control style

````python
from typing import Dict, Union

from sympy import re

@tool 
def sum_numbers_with_complex_output(inputs:str) -> Dict[str, Union[float, str]]:
    """
    Extract and sum all integers and decimals from the input
    """
    matches = re.findall(r'-?\d+(?:\.\d+)?', inputs)

    if not matches:
        return {"result": "No numbers found in the input."}
    
    try:
        numbers = [float(num) for num in matches]
        total = sum(numbers)
        return {"result": total}
    except Exception as e:
        return {"result": f"Error processing numbers: {str(e)}"}
    
# create a ReAct-style agent in LangGraph
# use the create_react_agent function

from langgarph.prebuilt import create_react_agent

add_agent = create_react_agent(
    tools=[sum_numbers_with_complex_output],
    llm = llm,
    prompt = "You are a helpful mathematical assistant that can perfomr various operatons. use the tools precisely and explain your resoning clearly.",
    verbose = True,
    handle_parsing_errors = True
)

response = add_agent.invoke({
    "messages": [("human", "Add the numbers -10, -20, -30)]")]
})

# Get the final answer
final_anser = response["messages"][-1].content
````

#### Orchestrating multiple tools to build a Math toolkit
- sometime a single tool is not enough
- some task require different tool
    - example: Banking transactions
        - one tool for Deposits
        - Withdrawals
        - Transfers


````python
from langchain_core.tools import tool
import re

@tool
def add_numbers(inputs:str) -> dict:
    """
    Adds a list of numbers
    """
    numbers = [int(num) for num in re.findall(r"\d+", inputs)]
    if not numbers:
        return {"result": "No numbers found in the input."}
    result = sum(numbers)
    return {"result": result}   

@tool
def multiply_numbers(inputs:str) -> dict:
    """
    Adds a list of numbers
    """
    numbers = [int(num) for num in re.findall(r"\d+", inputs)]
    if not numbers:
        return {"result": "No numbers found in the input."}
    result = 1
    for num in numbers:
        result *= num
    return {"result": result}


# create a ReAct-style agent in LangGraph
# use the create_react_agent function

from langgarph.prebuilt import create_react_agent

add_agent = create_react_agent(
    tools=[add_numbers, multiply_numbers],
    llm = llm,
    prompt = "You are a helpful mathematical assistant that can perfomr various operatons. use the tools precisely and explain your resoning clearly.",
    verbose = True,
    handle_parsing_errors = True
)

response = add_agent.invoke({
    "messages": [("human", "Add the numbers -10, -20, -30)]")]
})

# Get the final answer
final_anser = response["messages"][-1].content
````
