### A person can take 1 step or 2 step. How many ways he can take n steps

- logic
    - suppose 8 setp 
        - 1 step from 7th step
        - 2 step from 6th step

    - in that way f(n) = f(n-1)+f(n-2)

    ### Break down
    - 1st step - 1 way
    - 2nd step - 2 ways (1+1, or 2)
    - 3rd step - f(1) + f(2) = 3 ways
    - 4th step - f(2) + f(3) = 3 + 2 = 5 ways

```python
def climb_stairs(n):
    if n <= 2:
        return n
    
    prev_2 = 1
    prev_1 = 2
    for i in range(3, n + 1):
        current = prev_1 + prev_2
        prev_2 = prev_1
        prev_1 = current

    return current

print(climb_stairs(5))  # Output: 8
```

### 0-1 kanpsasck

```python
def knapsack_2d(weights, profits, capacity):
    n = len(weights)
    
    # Create a 2D grid initialized with zeros
    # Rows: items (0 to n)
    # Columns: capacities (0 to capacity)
    dp = [[0 for _ in range(capacity + 1)] for _ in range(n + 1)]
    
    # Build the table bottom-up
    for i in range(1, n + 1):
        for w in range(1, capacity + 1):
            current_weight = weights[i - 1]
            current_profit = profits[i - 1]
            
            if current_weight <= w:
                # Option 1: Take the item 
                # (current profit + best profit of the remaining weight)
                take_item = current_profit + dp[i - 1][w - current_weight]
                
                # Option 2: Leave the item
                leave_item = dp[i - 1][w]
                
                # Pick the maximum of the two options
                dp[i][w] = max(take_item, leave_item)
            else:
                # The item is too heavy for the current capacity, so leave it
                dp[i][w] = dp[i - 1][w]
                
    # The bottom-right cell contains the maximum profit
    return dp[n][capacity]

# --- Testing with our example ---
weights = [1, 3, 4]   # Item A, B, C weights
profits = [15, 20, 30] # Item A, B, C profits
capacity = 4

max_profit = knapsack_2d(weights, profits, capacity)
print(f"Maximum Profit: ${max_profit}")  # Outputs: Maximum Profit: $35
```
