### Core components of multi-agent systems
- agent: Autonomus units with specific capabilites and goals
- Environment: Context within which agents operate and interact
- Communication protocols: Standrs that enable agtens to share information

### Agent specilization concetps
- **Capability boundaries**: each agent should have a well-defined and focused scope
- **Exeprtise depth vs Breadth**:
    - need to distingush specilist and generalist agents
    - generalist are routing tasks and monitoring progress
- **Interface standardization**
    - communicate through structured inputs and outputs
    - often JSON
- **Handoff patterns**
    - pass tasks when outside their expertise to other agents
    - a document retreiver agent hadover the information to a summarization agent

### Advantates of multi-agnet systems
- scalability: Easily add or remove agents
- Flexibility: Adapt to changes in the enviromnet or task
- Robustness: System can continue functioning even if some agents fail
