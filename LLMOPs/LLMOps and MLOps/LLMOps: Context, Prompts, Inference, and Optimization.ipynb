## Context Windows
- **context windows** and efficient **tocken budget**
    - not more text but efficient text in the context window
    - context window is limited space, we need to use this efficiently

### Token Budget Fromula
- **Budget calculation**: Model_limit - system_prompt - ouput_reserve = chunk_to_fit
    - output_reserve: the context reserved for the ai generation
- **Recency bias**: Most relevent chunks go LAST in context (transformers give more importance to the recent information)
- **Metadata injection**: source: faq, date: 2023-12-01 before each chunk
- **Accurate counting**: use some tokenizer to count the exact count
    - use tokenizer like `tiktoken` 
    - never guss or overfit

### RAG Pipeline Example
- system prompt: 400 tokens, ouput reserve: 800 tokens -> 2800 for docs

    - Encode each retrieved chunk with ticktockn, sort by relevence score desc
    - Fill from most relevant down until cumilative tokens <= 2800
    - Inject metadata header per chunk: source, data, relevance_score
    - Most relevant chunk placed LAST - transformer recncy bias

### Multi-Turn Truncation
- Truncate oldest messages fist in multi-turn conversations.
- Preserve system prompt + last N turn always

### Key Principles
- **Budget first**: Calculate available space before retrieval
- **Recency matters**:  Place mst relevant content last for better attention
- **Use tiktoken**: Accurate token counting prevents truncation errors
- **Log usage**: Monitor toekns_used/total_budget ratio for optimization
