### Token Budget Fromula

    - output_reserve: the context reserved for the ai generation
- **Recency bias**: Most relevent chunks go LAST in context (transformers give more importance to the recent information)

- **Accurate counting**: use some tokenizer to count the exact count

    - never guss or overfit

### Prompt Architecture Patterns for Production LLM Systems

- Designing efficient context windows involves strategically ordering retrieved content, injecting only the most relevant metadata, and respecting token limits to maximize the useful information available to the LLM. Research shows that LLMs attend more strongly to content at the beginning and end of the context window (the "lost in the middle" effect), so placing the most relevant documents at these positions improves response quality. Metadata injection—adding source titles, dates, and relevance scores alongside retrieved text—helps the model ground its responses and enables source attribution. [Read more](https://redis.io/blog/llm-token-optimization-speed-up-apps/)

- Structured prompts that clearly separate system instructions, user context, retrieved documents, and output format specifications produce more reliable and deterministic outputs. Production prompt architectures typically use templating systems that enforce consistent structure across all invocations while allowing dynamic content injection. Techniques like XML-tagged sections, numbered document references, and explicit constraint statements reduce ambiguity and make prompts easier to version, test, and debug. [Read more](https://www.braintrust.dev/articles/what-is-prompt-versioning)

- Reducing hallucination through prompt engineering involves explicit grounding instructions ("answer only based on the provided documents"), confidence calibration ("say 'I don't know' if the context doesn't contain the answer"), and citation requirements that force the model to reference specific source passages. These techniques are complementary to retrieval optimization—better prompts reduce hallucination given fixed retrieval quality, while better retrieval reduces hallucination given fixed prompts. Combining both approaches produces the most reliable production systems. [Read more](https://redis.io/blog/llm-token-optimization-speed-up-apps/)
