## Multi-Modal Retrival-Augmented-Generation (MM-RAG)

### Learning Objectives


- MM-RAG
    - understand wide range of data
        - images
        - text etc
    - Retrieve infromation
    - Generate responses



### Why MM-RAG


### MM-RAG pattern
1. Mulitmodal data retrieval
    - used specilized retrievers
    - can process text doc, images, audio recordings, and videos
2. Contrastive learning
    - Train models that create related data
    - Helps system connect images and text easily
        - eg text "a domestic feline" is equal to a "cats picture"
3. Generative models informed by multimodel context
    - response based on rich array of infromation





## MM-RAG pipeline


- Data indexing
    - various data types are converted to index
        - text, images, audio, and video
    - the index are stored in a vector db
- Dara Retreival
    - User query
        - text, image, or both
    - converted to embeddings
    - serached in vector db accross all modalities
- Augmentation
    - combined retreived multimodal data with original user query
- Response generation
    - augmented query to multimodal generative model
    - produce response that ingegrate information from various retreived modalities
